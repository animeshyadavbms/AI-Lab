```python
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def depth_limited_search(state, goal, depth, path, on_path):
    if state == goal:
        return path

    if depth == 0:
        return None

    state_tuple = tuple(state)
    on_path.add(state_tuple)

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state_tuple = tuple(new_state)

            if new_state_tuple not in on_path:

                result = depth_limited_search(
                    new_state,
                    goal,
                    depth - 1,
                    path + [(move, new_state.copy())],
                    on_path
                )

                if result is not None:
                    return result

    on_path.remove(state_tuple)

    return None


def IDS(initial, goal):

    depth = 0

    while True:

        print("================================")
        print("Searching at depth:", depth)
        print("================================")

        result = depth_limited_search(
            initial,
            goal,
            depth,
            [],
            set()
        )

        if result is not None:
            return result

        depth += 1



initial = [
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
]

goal = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]


print("INITIAL STATE:")
print_state(initial)

print("GOAL STATE:")
print_state(goal)


solution = IDS(initial, goal)


print("================================")
print("SOLUTION")
print("================================")

print("Initial State:")
print_state(initial)


for move, state in solution:
    print("Move:", move)
    print_state(state)


print("Goal Reached!")
print("Number of moves:", len(solution))
```
