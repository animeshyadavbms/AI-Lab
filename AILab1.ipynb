```python
b = [" "] * 9

def show():
    print(f"{b[0]}|{b[1]}|{b[2]}")
    print("-+-+-")
    print(f"{b[3]}|{b[4]}|{b[5]}")
    print("-+-+-")
    print(f"{b[6]}|{b[7]}|{b[8]}")

win = [(0,1,2),(3,4,5),(6,7,8),
       (0,3,6),(1,4,7),(2,5,8),
       (0,4,8),(2,4,6)]

p = "X"

for _ in range(9):
    show()
    n = int(input(f"Player {p}, enter position (1-9): ")) - 1

    if b[n] != " ":
        print("Already occupied!")
        continue

    b[n] = p

    if any(b[a] == b[c] == b[d] == p for a,c,d in win):
        show()
        print(p, "wins!")
        break

    p = "O" if p == "X" else "X"
else:
    show()
    print("Draw!")
```

```python
import string

def get_room_names(num_rooms):
    """
    Generates room names like 'A', 'B', 'C', ... up to num_rooms.
    Limits to 26 rooms (A-Z) for simplicity of naming.
    """
    if num_rooms <= 0:
        raise ValueError("Number of rooms must be positive.")
    if num_rooms > 26:
        print("Warning: Limiting to 26 rooms (A-Z) for simplicity of naming.")
        num_rooms = 26
    return [string.ascii_uppercase[i] for i in range(num_rooms)]

def reflex(location, status, room_names_list):
    """
    Reflex agent for N rooms.
    If the current room is dirty, it sucks.
    If clean, it moves to the next room in a circular fashion within the room_names_list.
    """
    if status == "Dirty":
        return "Suck"
    else:
        current_index = room_names_list.index(location)
        next_index = (current_index + 1) % len(room_names_list)
        return f"Move_To_{room_names_list[next_index]}"

def goal_agent(room_states, current_location, room_names_list):
    """
    Goal-based agent for N rooms.
    If the current room is dirty, it sucks.
    Otherwise, it finds the first dirty room (in order of room_names_list) and moves towards it.
    If all rooms are clean, it achieves its goal.
    """
    if room_states[current_location] == "Dirty":
        return "Suck"

    for room in room_names_list:
        if room_states[room] == "Dirty":
            return f"Move_To_{room}"

    return "Goal Achieved"

while True:
    try:
        n_rooms = int(input("Enter the number of rooms (e.g., 3 for A, B, C): "))
        if n_rooms <= 0:
            print("Please enter a positive number of rooms.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter an integer.")

room_names = get_room_names(n_rooms)
print(f"Rooms generated: {', '.join(room_names)}")

print("\n--- Reflex Agent Test ---")
test_loc = ""
while test_loc not in room_names:
    test_loc = input(f"Location for reflex agent test {tuple(room_names)}: ").upper()
    if test_loc not in room_names:
        print(f"Invalid location. Please choose from {tuple(room_names)}.")

test_status = input(f"Status of room {test_loc} (Clean/Dirty) for reflex agent test: ").capitalize()

reflex_action = reflex(test_loc, test_status, room_names)
print(f"Reflex Action from {test_loc} ({test_status}): {reflex_action}")


print("\n--- Goal-Based Agent Simulation ---")

room_states = {name: "Dirty" for name in room_names}

loc = ""
while loc not in room_names:
    loc = input(f"Enter initial location for Goal-Based Agent {tuple(room_names)}: ").upper()
    if loc not in room_names:
        print(f"Invalid location. Please choose from {tuple(room_names)}.")

print("\n--- Starting simulation ---")
print("Initial room states:", room_states)
print("Initial location:", loc)

action_count = 0
while True:
    action_count += 1
    print(f"\nStep {action_count}: Current Location: {loc}, Room States: {room_states}")
    action = goal_agent(room_states, loc, room_names)
    print("Action:", action)

    if action == "Goal Achieved":
        print("All rooms are clean!")
        break
    elif action == "Suck":
        room_states[loc] = "Clean"
    elif action.startswith("Move_To_"):
        new_loc = action.split("_")[2]
        loc = new_loc
    else:
        print(f"Error: Unknown action '{action}' from goal_agent.")
        break
```
